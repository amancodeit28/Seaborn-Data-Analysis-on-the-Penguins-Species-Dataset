# KDE Plot for estimating thr probablity density function of continuous or non-parametric data
# Or in simple terms: a smooth that shows the distribution of data - like a smoothed-out version
# It helps you see where the data is concentrated (peak) and how it spreads

```python
## KDE Plot
sns.kdeplot(data = penguins, x = "body_mass_g", hue = "species", fill = True)
```

## Pair Plot
# This is basically a grid of plots which shows all the relation between all the numerical variables in a DataFrame
# It helps you visualize multiple plots together

```python
sns.pairplot(data=penguins[columns])
```

```python
sns.pairplot(data = penguins, hue = "sex", palette = "Set2", diag_kind = "kde")
```

```python
sns.pairplot(data = penguins, hue = "sex", palette = "Set2", diag_kind = "hist")
```

```python
## Pair grid -> subplot for plotting relationships in a dataset

graph = sns.PairGrid(data = penguins, hue = "sex", palette = "Set2")
graph.map_upper(sns.scatterplot) #upper triangle
graph.map_lower(sns.kdeplot) #lower triangle
graph.map_diag(sns.histplot) #diagnol
graph.add_legend()
```
